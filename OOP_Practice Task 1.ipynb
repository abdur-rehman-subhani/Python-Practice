Created a Class Person with some attributes like name,age etc..

```python
class Person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def displayInfo(self):
        print(f'The name  is: {self.name}')
        print(f'The age is: {self.age}')
```

Created a class Student that inherits from Class person. Student class also has its own attribute and an overridden method

```python
class Student(Person):
    def __init__(self, name, age,id):
        super().__init__(name, age)
        self.student_Id=id
    def displayInfo(self):
        super().displayInfo()
        print(f"The student Id of {self.name} is: {self.student_Id}")
```

Created objects of both classes and run method displayInfo() on both the objects to create some output

```python
person=Person('Harry',24)
person.displayInfo()

student=Student("David",29,6732)
student.displayInfo()
```
