Created a class Shape with a method area() in it.

```python
import math
class Shape:
    def __init__(self):
        pass
    def area(self):
        return 0
```

Created a class Rectangle that inherits from Shape with its own attributes and an overridden method area() providing it's own implementation

```python
class Rectangle(Shape):
    def __init__(self,width,height):
        super().__init__()
        self.width=float(width)
        self.height=float(height)
    def area(self):
        return (self.height)*(self.width)
```

Created a class Circle that inherits from Shape with its own attributes and an overridden method area() providing it's own implementation

```python
class Circle(Shape):
    def __init__(self,radius):
        super().__init__()
        self.radius=float(radius)

    def area(self):
        return math.pi *(self.radius**2)
      
```

Created objects of Circle class and Rectangle Class and run the area() method to create some output

```python
circle=Circle(6.6)
print(f"Area of Circle: {circle.area()}")
rectangle=Rectangle(2.4,6)
print(f"Area of Rectangle: {rectangle.area()}")
```
