Created a base class Animal with a method speak()

```python
class Animal:
    def __init__(self):
        pass
    def speak(self):
        print("The animal is speaking.....!")
```

Created a class Bird that inherits from Animal with a overridden method speak() to provide its own implementation

```python
class Bird(Animal):
    def __init__(self,sound):
        super().__init__()
        self.sound=sound
    def speak(self):
        print(f"The Bird is speaking \'{self.sound}\'")
```

Wrote a function make_animal_speak() that takes an Animal object and calls its speak()
method, demonstrating polymorphism

```python
animal=Animal()
def make_animal_speak(animal):
    animal.speak()

make_animal_speak(animal)
```
